# Veri Okuryazarlıği nedir?
- Veri okuryazarlığı her turden veri tipini degisken ve olcek turlerini tanımlayabilme, betimsel, istatikleri ve istatiksel grafikleri kullanarak ver ideğerlendirebilme yeteneğidir.
- Veri ile ilk temas ettigimizde veriyi değerlendirme yorum yapabilme yetenegidir.

# Populasyon ve Örneklem
- Populasyon ilgilendigimiz hedef kitlemizdir.
- Örneklem bu populasyon içinden secilen bir alt kümedir.
- Veriden faydaklı bilgi çıkarmak icin veriti temsil eden alt kumeye örneklem denir.
- Bu örneklem ana kümeyi populasyonu iyi temsil etmesidir ve yansız olmalıdır.

# Gözlem birimi
- Örneklem (alt küme icindeki) deki her bir eleman gözlem birimi olarak adlandırılır. Araştırmanın ana odak noktasıdır.

# Değişken ve Değişken Türleri
- Değişken birimden birime farklı değer alan niceliklerdir.

## Değişken TÜrleri
* Sayısal değişkenler (nicel, kantitatif)
* Kategorik değişkenler (nitel, kalitatif) --> Evet hayır, cinsiyet vb kategorik değişkenler

## Ölçek Türleri
- Sayısal değişkenler için : Aralık ve Oran
 - başlangıc noktası sıfır olmayan sayısal değişkenlerin ölçek türü aralıktır.
 - Başlangıc noktası sıfır kabul edilen sayısal değişkenlerin ölçek türü orandır.
- Kategorik değişkenler için: Nominal ve Ordinal. Karakterlerden olusan, Text formatında string olarak geçiyor.
- Nominal : Cinsiyet kategorik değişkendir. kadın ve erkek is bu kategorik değişkenin sınıflarıdır. Kadın ve Erkek sınıfları arasında fark olmadığı için bu değişken nominal ölçek türüne sahiptir.
- Ordinal : Sınıfları arasında fark olurması durumudur. Rütbe kategorik bir değişkendir. Bu kategorik değişkenin sınıfları
- Onbaşı < Yüzbaşı < Binbaşı < Albay
- Değişken sınıfları arasında fark olduğu için Rütbe değişkeni ordinaldir.

## Medyan
Bir seriyi küçükten büyüğe veya büyükten küçüge doğru sıraladığımızda tam orta noktadan seriyi iki eşit parçaya ayıran değere medyan denir. Elimizde 5 deger ise 3 deger ortada ama 6 deger varsa ortada kalan sayıların ortalamasını alıyoruz.
Seriyi kucukten buyuge dogru sıralıyoruz
10 12 13 15 17 o zamana medyan 13
10 12 13 13 15 17 ise (13+13)/2 = 13 olarak medyanı buluruz

Medyan ve ortalama karsılastırılması. Eger verimiz simetrik değilse yanı aykırı degerlr varsa  ortalama hesaplama yanıltıcı olacaktır.

# Dağılım Ölçüleri
1. Değişim Aralığı = Maksimum değer - Minimum değer
Değişkenlerin hangi arlaıkta dagıldıgına bakıyoruz. Örnegin maasların değişim aralıgı kucuk olan yerlerde maasların yakın oldugu kanısına varılabilir bu duruma gore yorum yapılabilir.

2. Standart sapma : ortalamadan olan sapmanın genel bir olcusu. Ortalamaya olan uzaklıgı hesaplıyoruz. Ortalamadan olan sapmaların ortalamasıdır.

3. Varyans : cok kapaca standart sapmanın karesidir.  birde nfazla degiskenin dagılımını kıyaslamak icin kullanılır.

4. Çarpıklık : bir degiskenin dagılımın simetrik olamayısıdır. pozitif (egrinin sola kayması) veya negaitif carpıklık (egrinin saga kayması) olarak degerlendiirlir. Tepe noktası tam ortada ise capıklık 0 olur, yani simetrik.

5. Basıklık : Degisken dagılımının sivriligini / basıklıgını gösteren istatisitktir.
