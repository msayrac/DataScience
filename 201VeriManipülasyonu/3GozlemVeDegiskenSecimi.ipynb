# Birleştirme (Join) İşlemleri

```python
import numpy as np
import pandas as pd

m = np.random.randint(1,30,size=(5,3))
df1 = pd.DataFrame(m,columns = ["var1","var2","var3"])
df1
```

```python
df2 = df1 + 99
df2
```

```python
pd.concat([df1,df2]) # index lerde 0 dan baslar her liste icin
```

```python
pd.concat([df1,df2],ignore_index=False)
```

```python
pd.concat([df1,df2],ignore_index=True)
```

# ileri birleştirme islemleri


```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({"calisanlar": ["Ali","Veli","Ayse","Fatma"],"grup":["Muhasebe","Muhendislik","Muhendislik","İK"]})
df1
```

```python
df2 = pd.DataFrame({"calisanlar":["Ayse","Ali","Veli","Fatma"],"ilk_giris":[2010,2009,2014,2019]})
df2
```

```python
pd.merge(df1,df2) # merge bire bir birleştirme default olarak calisanlar ortak oldugu icin calisanlara göre birleştirdi
```

```python
pd.merge(df1,df2, on="calisanlar")
```

```python
# coktan tekile many to one
df3 = pd.merge(df1,df2)
df3
```

```python
df4 = pd.DataFrame({"grup":["Muhasebe","Muhendislik","İK"],"mudur":["Caner","Mustafa","Berkcan"]})
df4
```

```python
pd.merge(df3,df4)
```

```python
# many to many coktan coka
df5 = pd.DataFrame({"grup":["Muhasebe","Muhasebe","Muhendislik","Muhendislik","İK","İK"],"yetenekler":["matematik","excel","kodlama","linux","excel","yonetim"]})
df5
```

```python
df1
```

```python
pd.merge(df1,df5) # her calısanın yeteneklerini belirtir
```
