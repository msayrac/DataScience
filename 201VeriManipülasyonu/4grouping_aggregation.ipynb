# Gruplama işlemleri

```python
import pandas as pd
df = pd.DataFrame({"gruplar":["A","B","C","A","B","C"],"veri":[10,11,52,23,43,55]},columns=["gruplar","veri"])
df
```

```python
df.groupby("gruplar") # grupları tutar
```

```python
df.groupby("gruplar").mean() # tutulan grupların ortalamasını al
```

```python
df.groupby("gruplar").sum()
```

```python
df.groupby("gruplar").count()
```

```python
# aggregate, filter, transform, apply --< ileri seviye veri analizinde kullanılan fonksiyonlar
import pandas as pd
df = pd.DataFrame({"gruplar":["A","B","C","A","B","C"],"degisken1":[10,23,33,22,11,99],"degisken2":[100,253,333,262,111,969]},columns=["gruplar","degisken1","degisken2"])
df
```

```python
# aggregate
df.groupby("gruplar").mean() # gruplara gore ortalama hesaplar. veri setini gruplara göre böl
```

```python
import numpy as np
df.groupby("gruplar").aggregate([min,np.median,max])
```

```python
df.groupby("gruplar").aggregate({"degisken1":"min","degisken2":"max"})
# gruplara göre böl aggregate ile degisken1 icin min degisken2 icin max bul
```

```python
# filter klasik filtrelemerde kosul calıstırırız. mevcut pandas fonk bu kosul yok ç fonk kendimiz tanımlamamız gerekiyor
import pandas as pd
df = pd.DataFrame({"gruplar":["A","B","C","A","B","C"],"degisken1":[10,23,33,22,11,99],"degisken2":[100,253,333,262,111,969]},columns=["gruplar","degisken1","degisken2"])
df
```

```python
def filter_func(x):
    return x["degisken1"].std() >9
```

```python
df.groupby("gruplar").std()
```

```python
df.groupby("gruplar").filter(filter_func)
```

```python
# transform desıskenlerde donusturme / degisiklik

df["degisken1"]*9
```

```python
df_a = df.iloc[:,1:3]
df_a
```

```python
# tanımlı bir fonk yoksa biz degisiklik yapabiliriz
df_a.transform(lambda x: (x-x.mean())/x.std()) #butun degiskenlerde degiisklik yapar
```

```python
# apply
import pandas as pd
df = pd.DataFrame({"gruplar":["A","B","C","A","B","C"],"degisken1":[10,23,33,22,11,99],"degisken2":[100,253,333,262,111,969]},columns=["gruplar","degisken1","degisken2"])
df
```

```python
df.groupby("gruplar").apply(np.sum)
```

```python
df.groupby("gruplar").apply(np.mean)
```
