# Pivot Table

```python
import pandas as pd
import numpy as np
import seaborn as sns

titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic.groupby("sex")["survived"].mean() # sag kalanların ortalaması
```

```python
titanic.groupby("sex")[["survived"]].mean() # sag kalanların ortalaması
```

```python
titanic.groupby(["sex","class"])[["survived"]].aggregate("mean")
# kadınlardan first second third class da yasayanlardan yuzde kacı kurtulmus
# benzer sekilde erkeklerden first second an dthird class yuzde kacı kurtulmus
```

```python
titanic.groupby(["sex","class"])[["survived"]].aggregate("mean").unstack() # unstack() okumayı daha kolay olmasını saglar
```

```python
# pivot ile pivot table

titanic.pivot_table("survived",index="sex",columns="class")
# odaklanan degisken ""survived
# index =sex  satır olarak degerlendiriyor
# class degiskenine de odaklan stunlara class olarak atanır
```

```python
age = pd.cut(titanic["age"],[0,18,90])
# titanic verisindeki age degiskenini bol. 0, 18,90 degerlerine göre bol
age.head(10)
```

```python
titanic.pivot_table("survived",["sex",age],"class")
```
