# Array Birlestirme (concatenation)

```python
import numpy as np
```

```python
x = np.array([1,2,3])
y = np.array([4,5,6])
```

```python
np.concatenate([x,y])
```

```python
z = np.array([7,8,9])
```

```python
z
```

```python
np.concatenate([x,y,z])
```

```python
# iki boyut
a = np.array([[1,2,3],
             [4,5,6]])
```

```python
np.concatenate([a,a], axis=0) # satır baz birlestirme
```

```python
# sütün baz birleştirme
np.concatenate([a,a],axis=1)
```
